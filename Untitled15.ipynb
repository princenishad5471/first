```python
import pandas as pd
import numpy as np
```

```python
car=pd.read_csv('/content/quikr_car.csv')
```

```python
backup=car.copy()
```

```python
car=car[car['year'].str.isnumeric()]
```

```python
car['year'] = car['year'].astype(int)
```

```python
car=car[car['Price']!='Ask For Price']
```

```python
car['Price']=car['Price'].str.replace(',','').astype(int)
```

```python
car['kms_driven']=car['kms_driven'].str.split(' ').str.get(0).str.replace(',','')
```

```python
car=car[car['kms_driven'].str.isnumeric()]
```

```python
car['kms_driven']=car['kms_driven'].astype(int)
```

```python
car=car[~car['fuel_type'].isna()]
```

```python
car['name']=car['name'].str.split(' ').str.slice(0,3).str.join(' ')
```

```python
car=car.reset_index(drop=True)
```

```python
car=car[car['Price']<6e6].reset_index(drop=True)
```

```python
car.to_csv('Cleaned_Car_data.csv')
```

# **Model**

```python
x=car.drop(columns='Price')
y=car['Price']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline


```

```python
ohe=OneHotEncoder()
ohe.fit(x[['name','company','fuel_type']])
```

```python
column_trans=make_column_transformer((OneHotEncoder(categories=ohe.categories_),['name','company','fuel_type']),remainder='passthrough')
```

```python
lr=LinearRegression()
```

```python
pipe=make_pipeline(column_trans,lr)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred=pipe.predict(x_test)

```

```python
r2_score(y_test,y_pred)
```

```python
scores=[]
for i in range(1000):
  x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=i)
  lr=LinearRegression()
  pipe=make_pipeline(column_trans,lr)
  pipe.fit(x_train,y_train)
  y_pred=pipe.predict(x_test)
  scores.append(r2_score(y_test,y_pred))
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=np.argmax(scores))
lr=LinearRegression()
pipe=make_pipeline(column_trans,lr)
pipe.fit(x_train,y_train)
y_pred=pipe.predict(x_test)
r2_score(y_test,y_pred)
```

```python
import pandas as pd
import pickle
from sklearn.linear_model import LinearRegression
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

# Load data from the already cleaned 'car' DataFrame
df = pd.read_csv('Cleaned_Car_data.csv') # Load cleaned data directly from CSV

X = df[['name', 'company', 'year', 'kms_driven', 'fuel_type']]
y = df['Price'] # 'Price' with capital 'P' as in the 'car' DataFrame

# Preprocessing
categorical_cols = ['name', 'company', 'fuel_type']
numeric_cols = ['year', 'kms_driven']

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
        ('num', 'passthrough', numeric_cols)
    ]
)

# Model
model = LinearRegression()

# Pipeline
pipe = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', model)
])

# Train
pipe.fit(X, y)

# Save model (IMPORTANT)
pickle.dump(pipe, open("LinearRegressionModel.pkl", "wb"))
```
