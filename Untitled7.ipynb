````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
````

````python
df=pd.read_csv('/content/compressed_data.csv.gz')
df
````

````python
df.columns
````

````python
df.describe()
````

````python
df.info()
````

````python
df.isnull().sum()
````

````python
df['last review']=pd.to_datetime(df['last review'],errors='coerce')
````

````python
df.fillna({ 'reviews per month':0,'last review':df['last review'].min()},inplace=True)
````

````python
df.dropna(subset=['NAME','host name'],inplace=True)
````

````python
df=df.drop(columns=['license','house_rules'],errors='ignore')
````

````python
df['price']=df['price'].replace('[\$,]','',regex=True).astype(float)
df['service fee']=df['service fee'].replace('[\$,]','',regex=True).astype(float)
````

````python
df.drop_duplicates(inplace=True)
````

````python
df.info()
````

````python
df.describe()
````

**Visualization**

# what is the distribution of listing price?


````python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=50,kde=True,color='red')
plt.title('Distribution of Listing Price')
plt.xlabel('Price $')
plt.ylabel('Frequency')
plt.show()
````



 **How are different room types distributed**

````python
plt.figure(figsize=(10,6))
sns.countplot(data=df,x='room type',palette='viridis')
plt.title('Distribution of Room Types')
plt.xlabel('Room Type')
plt.ylabel('Frequency')
plt.show()
````

# How are listing distributed across different neighborhoods?

````python
plt.figure(figsize=(12, 8))
sns.countplot(y='neighbourhood group',data=df,color='lightgreen',order=df['neighbourhood group'].value_counts().index)
plt.title('Distribution of Listings by Neighborhoods group')
plt.xlabel('count')
plt.ylabel('Neighborhoods group')
plt.show()
````

# What is the relationship b/w price and room type

````python
plt.figure(figsize=(10,6))
sns.boxplot(x='room type',y='price',hue='room type',data=df,palette='viridis')
plt.title('Price Vs Room Type')
plt.xlabel('Room Type')
plt.ylabel('Price $')
plt.legend(title='room type')
plt.show()
````

# How has the number of reviews change over with time

````python
df['last review']=pd.to_datetime(df['last review'])
reviews_over_time=df.groupby(df['last review'].dt.to_period('M')).size()

plt.figure(figsize=(12,6))
reviews_over_time.plot(kind='line',color='blue')
plt.title('Number of Reviews Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Reviews')
plt.show()
````
